```python
import torchxrayvision as xrv
import skimage
import torch
import torchvision

from PIL import Image

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import os
```

```python
import src.image_utils as image_utils
```

```python
from torchxrayvision.datasets import (
    NIH_Dataset,
    PC_Dataset
)
```

```python
models = [
    "densenet121-res224-all", 
    "densenet121-res224-rsna",
    "densenet121-res224-nih",
    "densenet121-res224-pc",
    "densenet121-res224-chex",
    "densenet121-res224-mimic_nb",
    "densenet121-res224-mimic_ch"
]

data_dir = str(repo_root / 'data')
```

```python
list((repo_root / 'data' / 'NIH').iterdir())
```

```python
nih_dataset = NIH_Dataset(
    os.path.join(data_dir, 'NIH', 'images-224')
)
```

```python
nih_dataset.labels.shape
```

```python
nih_dataset.csv
```

```python
model_choice = models[1]
model = xrv.models.DenseNet(weights=model_choice)
model.eval()

predictions = []
for _, row in nih_dataset.csv.iterrows():
    filename = os.path.join(data_dir, 'NIH', 'images-224', row['Image Index'])
    with Image.open(filename) as image:
        img = np.array(image.convert('RGB'))
    predicted_label = image_utils.predict(img, model=model)
    predictions.append({'image': row['Image Index'], **predicted_label})

pd.DataFrame(predictions)
```
